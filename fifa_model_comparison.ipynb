```python
import pandas as pd
from google.colab import files
from sklearn.model_selection import cross_val_predict, KFold
from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.metrics import mean_absolute_error

# Upload your dataset
uploaded = files.upload()
filename = next(iter(uploaded))

# Read CSV or Excel
if filename.lower().endswith(".csv"):
    df = pd.read_csv(filename)
else:
    df = pd.read_excel(filename)

# Select the eight predictors and target
X = df[[
    "fifa_rank_diff",
    "Elo_rank_diff",
    "Qualifying_gd_gap",
    "marketVal_diff",
    "Rest_days_diff",
    "goal_diff_history",
    "avg_age_diff",
    "squad_caps_diff"
]]

y = df["Goal Difference"]

# Compare three prediction methods
models = {
    "Mean Baseline": DummyRegressor(),
    "Linear Regression": LinearRegression(),
    "Ridge Regression": Ridge()
}

cv = KFold(n_splits=5, shuffle=True, random_state=42)
results = []

for name, model in models.items():
    predictions = cross_val_predict(model, X, y, cv=cv)
    error = mean_absolute_error(y, predictions)
    results.append({"Model": name, "MAE": round(error, 3)})

# Show best result first
results = pd.DataFrame(results).sort_values("MAE")

print("\nMODEL COMPARISON")
print(results.to_string(index=False))
print("\nLower MAE means more accurate predictions.")
```
